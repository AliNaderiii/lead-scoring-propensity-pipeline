# 01 - Exploratory Data Analysis & Data Quality
Real EDA on synthetic lead data (12K rows, 12% conversion)


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('../data/leads_raw.csv')
df.head()
print(f"Shape: {df.shape}, Conversion: {df['converted'].mean():.2%}")
df.describe()
```

```python
# Missing values
print(df.isna().sum())
print(f"Duplicates: {df.duplicated(subset=['lead_id']).sum()}")
# Class imbalance
sns.countplot(x='converted', data=df)
plt.title('Class Imbalance - 12% Conversion')
plt.show()
```

```python
# Feature vs target
import numpy as np
numeric_cols = df.select_dtypes(include=[np.number]).columns.drop('converted')
for col in ['engagement_score','credit_score','income','website_visits']:
    plt.figure()
    sns.boxplot(x='converted', y=col, data=df)
    plt.title(f'{col} vs Converted')
    plt.show()
```
